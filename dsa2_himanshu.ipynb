## 🔄 Code Segment

    prev = curr          # 3. Move 'prev' forward

Imagine the list is:

```
1 -> 2 -> 3 -> 4 -> None
```

- **Initial state:**
  - `prev = None`
  - `curr = head` (points to 1)

---

### Iteration 1
- `nxt = curr.next` → `nxt = 2`
- `curr.next = prev` → `1.next = None`  
  (we flip the arrow, so 1 points to `None`)
- `prev = curr` → `prev = 1`
- `curr = nxt` → `curr = 2`

Now the list looks like:
```
1 -> None
2 -> 3 -> 4 -> None
```

- `nxt = 3`
- `curr.next = prev` → `2.next = 1`
- `prev = 2`
- `curr = 3`

We’re **walking through the list** and flipping each arrow one by one.

````python
''' Structure of Linked List Node
class Node:
    def __init__(self, data): 
        self.data = data
        self.next = None
'''
from typing import Optional
class Solution:
    def detectCycle(self, head: Optional[Node]) -> Optional[Node]:
        slow = fast = head
        if head is None or head.next is None: 
            return None

        # Step 1: Detect if a cycle exists
        while fast and fast.next:
            fast = fast.next.next
            slow = slow.next
            if fast == slow:
                break
        else:
            # No cycle found
            return None

        # Step 2: Find the cycle entry point
        t1, t2 = head, slow
        while t1 != t2:
            t1 = t1.next
            t2 = t2.next
        return t1

    def lengthOfLoop(self, head: Optional[Node]) -> int:
        start = self.detectCycle(head)
        if start is None:
            return 0

        # Step 3: Count loop length
        count = 1
        current = start.next
        while current != start:
            current = current.next
            count += 1
        return count
        
        
````

You are given the head of a singly linked-list. The list can be represented as:

L0 → L1 → … → Ln - 1 → Ln
Reorder the list to be on the following form:

L0 → Ln → L1 → Ln - 1 → L2 → Ln - 2 → …
You may not modify the values in the list's nodes. Only nodes themselves may be changed.

 

Example 1:


Input: head = [1,2,3,4]
Output: [1,4,2,3]
Example 2:


Input: head = [1,2,3,4,5]
Output: [1,5,2,4,3]
 

Constraints:

The number of nodes in the list is in the range [1, 5 * 104].
1 <= Node.val <= 1000

````python
 # Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def reverse(self, head: Optional[ListNode]) -> Optional[ListNode]:
        curr = head
        prev = None

        while curr is not None:
            future = curr.next
            curr.next = prev
            prev= curr
            curr= future
        return prev
    def middle(self, head: Optional[ListNode]) -> Optional[ListNode]:
        slow = head
        fast = head

        while fast.next is not None and fast.next.next is not None:
            slow = slow.next
            fast = fast.next.next

        return slow
    def reorderList(self, head: ListNode | None) -> None:
        """
        Do not return anything, modify head in-place instead.
        """
        
        middleNode = self.middle(head)

        ll2 = middleNode.next
        middleNode.next = None

        ll2 = self.reverse(ll2)

        ll1 = head

        while ll1 is not None and ll2 is not None:

            f1 = ll1.next
            f2 = ll2.next

            ll1.next = ll2
            ll2.next = f1

            ll1 = f1
            ll2 = f2
````
