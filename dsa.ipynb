303. Range Sum Query - Immutable
Given an integer array nums, handle multiple queries of the following type:

Calculate the sum of the elements of nums between indices left and right inclusive where left <= right.
Implement the NumArray class:

NumArray(int[] nums) Initializes the object with the integer array nums.
int sumRange(int left, int right) Returns the sum of the elements of nums between indices left and right inclusive (i.e. nums[left] + nums[left + 1] + ... + nums[right]).
 

Example 1:

Input
["NumArray", "sumRange", "sumRange", "sumRange"]
[[[-2, 0, 3, -5, 2, -1]], [0, 2], [2, 5], [0, 5]]
Output
[null, 1, -1, -3]

Explanation
NumArray numArray = new NumArray([-2, 0, 3, -5, 2, -1]);
numArray.sumRange(0, 2); // return (-2) + 0 + 3 = 1
numArray.sumRange(2, 5); // return 3 + (-5) + 2 + (-1) = -1
numArray.sumRange(0, 5); // return (-2) + 0 + 3 + (-5) + 2 + (-1) = -3
 

Constraints:

1 <= nums.length <= 104
-105 <= nums[i] <= 105
0 <= left <= right < nums.length
At most 104 calls will be made to sumRange.

```python
from typing import List

class NumArray:

    def __init__(self, nums: List[int]):
        self.prefix = [0]*len(nums) # creating an array with 0 for the required size len(nums)
        self.prefix[0]=nums[0]
        for i in range (1,len(nums)):
            self.prefix[i]=self.prefix[i-1]+nums[i]


        

    def sumRange(self, left: int, right: int) -> int:
        # [-2, 0, 3, -5, 2, -1]
        # Prefix Array [-2,-2,1,-4,-2,-3]
        if left ==0:
            return self.prefix[right]
        else :
            return self.prefix[right] - self.prefix[left - 1]




nums = [-2, 0, 3, -5, 2, -1]
obj = NumArray(nums)
print(obj.sumRange(0,2))
print(obj.sumRange(2,5))
print(obj.sumRange(0,5))


# Time Complexity is O(n) --> creating the prefix Array
# O(1) --> getting the Sum
```

724. Find Pivot Index

Given an array of integers nums, calculate the pivot index of this array.

The pivot index is the index where the sum of all the numbers strictly to the left of the index is equal to the sum of all the numbers strictly to the index's right.

If the index is on the left edge of the array, then the left sum is 0 because there are no elements to the left. This also applies to the right edge of the array.

Return the leftmost pivot index. If no such index exists, return -1.

 

Example 1:

Input: nums = [1,7,3,6,5,6]
Output: 3
Explanation:
The pivot index is 3.
Left sum = nums[0] + nums[1] + nums[2] = 1 + 7 + 3 = 11
Right sum = nums[4] + nums[5] = 5 + 6 = 11
Example 2:

Input: nums = [1,2,3]
Output: -1
Explanation:
There is no index that satisfies the conditions in the problem statement.
Example 3:

Input: nums = [2,1,-1]
Output: 0
Explanation:
The pivot index is 0.
Left sum = 0 (no elements to the left of index 0)
Right sum = nums[1] + nums[2] = 1 + -1 = 0
 

Constraints:

1 <= nums.length <= 104
-1000 <= nums[i] <= 1000
 

Note: This question is the same as 1991: https://leetcode.com/problems/find-the-middle-index-in-array/

```python
from typing import List

class Solution:
    def pivotIndex(self, nums: List[int]) -> int:
        if not nums:
            return -1
            
        # Create and populate the prefix sum array
        prefix = [0] * len(nums)
        prefix[0] = nums[0]
        for i in range(1, len(nums)):
            prefix[i] = prefix[i-1] + nums[i]
            
        total_sum = prefix[-1]
        
        # Check for the pivot index
        for i in range(0, len(nums)):
            # Handle left sum for the 0th index gracefully
            left_sum = prefix[i-1] if i > 0 else 0
            right_sum = total_sum - prefix[i]
            
            if left_sum == right_sum:
                print("pivot index = " + str(i))
                return i
                
        return -1

Sol = Solution()
Sol.pivotIndex([1,7,3,6,5,6])
Sol.pivotIndex([1,2,3])
Sol.pivotIndex( [2,1,-1])
```

Prefix Sum --> L to R Sum
Suffix Sum --> R to L Sum

if you want the fastest, cleanest execution in Python, you can use itertools.accumulate to handle the prefix sums natively. This is often faster in practice because the loop runs in optimized C under the hood.

```python
from typing import List
import itertools

class Solution:
    def pivotIndex(self, nums: List[int]) -> int:
        total_sum = sum(nums)
        print(f"total is {total_sum}")
        
        # itertools.accumulate gives a running total: [nums[0], nums[0]+nums[1], ...]
        # Specifying initial=0 starts the accumulation at 0
        for i, left_sum in enumerate(itertools.accumulate(nums, initial=0)): # Returns (i=0,left_sum=1), (i=1,left_sum=3), (i=2,left_sum=6) 
            if i < len(nums) and left_sum == (total_sum - left_sum - nums[i]):
                print("pivot index = " + str(i))
                return i
        print("pivot index = -1")        
        return -1

Sol = Solution()
Sol.pivotIndex([1,7,3,6,5,6])
Sol.pivotIndex([1,2,3])
Sol.pivotIndex( [2,1,-1])
```

Leetcode 15: 3 Sum  **Important**
Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] such that i != j, i != k, and j != k, and nums[i] + nums[j] + nums[k] == 0.

Notice that the solution set must not contain duplicate triplets.

 

Example 1:

Input: nums = [-1,0,1,2,-1,-4]
Output: [[-1,-1,2],[-1,0,1]]
Explanation: 
nums[0] + nums[1] + nums[2] = (-1) + 0 + 1 = 0.
nums[1] + nums[2] + nums[4] = 0 + 1 + (-1) = 0.
nums[0] + nums[3] + nums[4] = (-1) + 2 + (-1) = 0.
The distinct triplets are [-1,0,1] and [-1,-1,2].
Notice that the order of the output and the order of the triplets does not matter.
Example 2:

Input: nums = [0,1,1]
Output: []
Explanation: The only possible triplet does not sum up to 0.
Example 3:

Input: nums = [0,0,0]
Output: [[0,0,0]]
Explanation: The only possible triplet sums up to 0.
 

Constraints:

3 <= nums.length <= 3000
-105 <= nums[i] <= 105

```python
class Solution:
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        nums.sort()
        res=[]
        for i in range(len(nums)-2):
            left = i+1
            right = len(nums)-1
            while left<right:
                total=nums[i]+nums[left]+nums[right]
                if total == 0:
                    triplet = [nums[i], nums[left], nums[right]]
                    if triplet not in res:
                        res.append(triplet)
                    left+=1
                    right-=1

                    while left < right and nums[left] == nums[left-1]:
                        left+=1
                    while left<right and nums[right]== nums[right+1]:
                        right-=1
                elif total>0:
                    right-=1
                else: 
                    left+=1
        return res

    
# complexity:
# Sort -> n log n
# nested while loop -> n^2
# so total complexity is n log (n) + n^2




obj = Solution()
print(obj.threeSum([-2,-1,-1,0,0,1]))
print(obj.threeSum([-1,0,1,2,-1,-4]))

        
```

You may assume that each input would have exactly one solution, and you may not use the same element twice.

You can return the answer in any order.

 

Example 1:

Example 2:

Input: nums = [3,2,4], target = 6

Example 3:

Input: nums = [3,3], target = 6
Output: [0,1]
 

Constraints:

2 <= nums.length <= 104
-109 <= nums[i] <= 109
-109 <= target <= 109
Only one valid answer exists.
 

Follow-up: Can you come up with an algorithm that is less than O(n2) time complexity?

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        hashmap = {}

        for i, num in enumerate(nums):
            complement = target - num

            if complement in hashmap:
                return [hashmap[complement], i]

            hashmap[num] = i
            print (hashmap[num])

        return []

obj = Solution()
obj.twoSum([1, 2, 3, 4], 6)
```

The majority element is the element that appears more than ⌊n / 2⌋ times. You may assume that the majority element always exists in the array.

 

Example 1:

Input: nums = [3,2,3]
Output: 3
Example 2:

Input: nums = [2,2,1,1,1,2,2]
Output: 2
 

Constraints:

n == nums.length
1 <= n <= 5 * 104
-109 <= nums[i] <= 109
The input is generated such that a majority element will exist in the array.
 

Follow-up: Could you solve the problem in linear time and in O(1) space?

```python
class Solution:
    def majorityElement(self, nums: list[int]) -> int:
        hashmap={}
        thr = len(nums)/2
        for i in nums:
            if i in hashmap:
                hashmap[i]+=1
            else:
                hashmap[i] = 1
            if hashmap[i] > thr :
                print(hashmap[i])
                return i
                
            # print(hashmap)
            

obj = Solution()
obj.majorityElement([1,1,1,1,2,2,2,2,2,2,3])
        
```

Sliding Window

[2,1,5,1,3,2]
Find the max sum of 3 consecutive elements

The idea is remove one element and add a new element and slide in this way



```python
def slide (nums: list[int]):
    sum = nums[0]+nums[1]+nums[2]
    max = sum
    for i in range(len(nums)-3):
        sum -= nums[i]
        sum += nums[i+3]
        if (sum > max):
            max = sum
        # print (sum, max)
    print (max)

slide ([2,1,5,1,3,2])
```

Get max avg of continuous k elements

```python
def findMaxAverage(self, nums: list[int], k: int) -> float:
        sum = 0
        for i in range (k):
            sum +=nums[i]
        avg = sum/k

        for i in range (len(nums)-k):
            sum -= nums[i]
            sum += nums[i+k]
            new_avg = sum/k
            if (avg > new_avg):
                avg = new_avg
        return avg

findMaxAverage ([1,12,-5,-6,50,3],4)
```
