```python
def binarysearch(nums: list[int], key: int) -> int:
    s, e = 0, len(nums) - 1
    
    while s <= e:
        m = (s + e) // 2
        if nums[m] < key:
            s = m + 1
        elif nums[m] > key:
            e = m - 1
        else:
            return m
    return -1

print(binarysearch([1, 2, 3, 4, 5, 6], 4))
```
