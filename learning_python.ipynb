```python
class p1:
    def p_1(self):
        print("p1")
    def test(self):
        print("test p1")

class p2:
    def p_2(self):
        print("p2")
    def test(self):
            print("test p2")

class c1(p1,p2): #Multiple inheritance
    def c_1(self):
        print("c1")

c = c1()
c.p_1()
c.p_2()
c.c_1()
c.test() 
# test p1
# Will call test from p1 
# It will simply take the first parent written in class c1(*p1*,p2) above
print(c1.__mro__)
# Or
print(c1.mro())

# (<class '__main__.c1'>, <class '__main__.p1'>, <class '__main__.p2'>, <class 'object'>)
# [<class '__main__.c1'>, <class '__main__.p1'>, <class '__main__.p2'>, <class 'object'>]
```

```python
class p1:
    def test(self):
        print("test p1")
        super().test()   # Calls next in MRO

class p2:
    def test(self):
        print("test p2")

class c1(p1, p2):
    def test(self):
        print("test c1")
        super().test()

c = c1()
c.test()
# test c1
# test p1
# test p2

# here we get both test p1 and test p2 because we use super()

# Unlike single inheritance, super() doesn’t just call the immediate parent — it follows the MRO chain.
```

```python
class A:
    def test(self):
        print("A")

class B(A):
    def test(self): #overriding
        print("B")
        super().test()

class C(A):
    def test(self):  #overriding
        print("C")
        super().test()

class D(B, C):
    def test(self):  #overriding
        print("D")
        super().test()

d = D()
d.test()
print(D.mro())
# D
# B
# C
# A
# [<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>]
```

Encapsulation

```python
class BankAccount:
    def __init__(self, balance):
        if balance < 0:
            raise ValueError("Balance cannot be negative")
        self.__balance = balance   # private field

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self.__balance += amount

    def withdraw(self, amount):
        if amount > self.__balance:
            raise ValueError("Insufficient funds")
        self.__balance -= amount

    def get_balance(self):
        return self.__balance

# Usage
account = BankAccount(100)
account.deposit(50)
account.withdraw(30)
print(account.get_balance())  # ✅ Always valid
```

1. Protecting Invariants
•	Invariants are rules that must always hold true.
•	In your class, the invariant is: balance can never be negative.
•	You enforce this by:
o	Checking in __init__ that the starting balance is ≥ 0.
o	Validating deposits (amount > 0).
o	Validating withdrawals (amount ≤ balance).
👉 This ensures the account object can never exist in an invalid state.
2. Immutable Objects
•	Your current class allows balance changes.
•	To make it immutable, you’d return a new account object instead of modifying the existing one:



```python
class ImmutableBankAccount:
    def __init__(self, balance):
        if balance < 0:
            raise ValueError("Balance cannot be negative")
        self.__balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        return ImmutableBankAccount(self.__balance + amount)

    def withdraw(self, amount):
        if amount > self.__balance:
            raise ValueError("Insufficient funds")
        return ImmutableBankAccount(self.__balance - amount)

    def get_balance(self):
        return self.__balance
account = BankAccount(100)
print(account.get_balance())   # Direct access to '_BankAccount__balance' is not allowed
print(account.__balance) 
```

👉 Now, every operation produces a new account, so the original object is never corrupted.

3. Getter/Setter Trap
Your get_balance() is fine, but imagine if you had a set_balance() method.

That would allow external code to break invariants (set_balance(-1000)).

Instead, you expose behavioral methods (deposit, withdraw) that enforce rules internally.

👉 This is why your design avoids the getter/setter trap.

4. Law of Least Knowledge (Demeter’s Law)
Suppose external code had to do:

python
account.get_balance() - withdrawal_amount
That’s exposing too much detail.

Better: let the account handle it internally with withdraw(amount).

External code only interacts with one method, not multiple internals.

👉 Your class already follows Demeter’s Law by keeping interactions short and safe.

5. API Design & Trust
Your API is minimal and intention-revealing:

deposit(amount) → clear purpose.

withdraw(amount) → enforces rules.

get_balance() → safe read-only access.

Developers can trust that using these methods won’t break the account’s integrity.

👉 This is the essence of encapsulation: safe, predictable APIs.

🔥 Advanced Twist: Distributed Systems
If this account were part of a banking microservice, you’d need to validate requests before persisting them.

Encapsulation ensures that even if external services send bad data, your domain model (BankAccount) rejects it.

✅ So your BankAccount example already demonstrates protecting invariants, avoiding setter traps, following Demeter’s Law, and building a trustworthy API.

🔍 How _ and __ work in Python
Single underscore (_var): A convention meaning “internal use only.” It does not enforce privacy. You can still access it directly.

Double underscore (__var): Triggers name mangling. Python renames it internally to _ClassName__var.

Example: self.__balance becomes self._BankAccount__balance.

This prevents accidental access, but you can still reach it if you know the mangled name.

👉 So by default, neither _ nor __ will throw an error. They just make access harder or discouraged.

⚡ How to Force Errors
If you want _ or __ to truly cause errors when accessed directly, you need to override attribute access:

1. Using __getattr__ or __getattribute__
You can intercept attribute access and raise errors if someone tries to touch _ or __ fields.

```python
class BankAccount:
    def __init__(self, balance):
        if balance < 0:
            raise ValueError("Balance cannot be negative")
        self.__balance = balance

    def __getattribute__(self, name):
        if name.startswith("_"):
            raise AttributeError(f"Direct access to '{name}' is not allowed")
        return super().__getattribute__(name)

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self.__balance += amount

    def withdraw(self, amount):
        if amount > self.__balance:
            raise ValueError("Insufficient funds")
        self.__balance -= amount

    def get_balance(self):
        return self.__balance

account = BankAccount(100)
print(account.get_balance())   # Direct access to '_BankAccount__balance' is not allowed
print(account.__balance)       # ❌ AttributeError: Direct access not allowed
```

```python
# In Python – Access Specifiers mean nothing. Everything is always public, unlike Java/C++.
# Starting with _ = Protected Member = Meant for internal use only.
# Starting with __ = Private. Access is blocked. But if assigned, a new copy is created in the memory
# maintains two copies as soon as an assignment happens

class BankAccount:

    def __init__(self, balance) -> None:
        self.__balance = balance
        self._bal = balance

    def deposit(self, amount):
        self.__balance += amount
        self._bal += amount

    # Getter
    def get_balance(self):
        return self.__balance
    
    def get_bal(self):
            return self._bal

account = BankAccount(1000)
account.deposit(9000)  
# print(account.__balance) # Error
account.__balance = 50  
# updates only here and not in class
# Creates a new member and not the one in class
print(account.get_balance()) #10000 getting from class
print(account.__balance) #50 getting from the prev line

print(account._bal) #10000
# No Error because '_' and not '__'

account._bal = 10 
#updates both in class and here
print(account.get_bal()) #10 
print(account._bal) #10

```

```python
from abc import ABC, abstractmethod

class Engine(ABC):
    def __init__(self, fuel_type):
        self.fuel_type = fuel_type

    @abstractmethod
    def start(self):
        print("Common startup logic")

    def stop(self):
        print("Engine stopped")


class PetrolEngine(Engine):
    def start(self):
        super().start()
        print("Petrol engine started")


engine = PetrolEngine("petrol")
engine.start()
engine.stop()
print(PetrolEngine.mro())

# Common startup logic
# Petrol engine started
# Engine stopped
# [<class '__main__.PetrolEngine'>, <class '__main__.Engine'>, <class 'abc.ABC'>, <class 'object'>]
```
